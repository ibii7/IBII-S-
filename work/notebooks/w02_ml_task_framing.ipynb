### 5. Why ML Beats a Fixed Rule
- **Complex Feature Interactions:** A simple rule like `IF days_inactive > 14 THEN churn` misses key nuance. A user with 14 days of inactivity who opened 5 customer support emails is very different from one who went silent entirely. ML captures non-linear relationships across many features simultaneously.
- **Adaptability to Concept Drift:** User behavior shifts as product features change. Fixed rules become outdated quickly and require manual maintenance, whereas an ML pipeline dynamically updates decision boundaries when retrained on new usage data.
- **Actionability:** Instead of a rigid yes/no rule, scoring gives each user a risk probability ($0.00 - 1.00$), enabling tiered retention actions (e.g., automated email for low-risk vs. human outreach for high-risk).
